```python
"""
product/scripts/03_build_connectivity_dataset.py

Builds a connectivity-based dataset from raw fMRI time series.
Computes pairwise ROI correlations using the CC200 atlas and saves
the result as a feature matrix with participant labels.

All arguments are optional. Run without arguments to process all
available participants, then use arguments to filter or subset as needed.

Usage:
    python 03_build_connectivity_dataset.py [--prefix] [--subdir]
                                            [--sex] [--age-min] [--age-max]
                                            [--seed]
                                            [--match-to-female-metadata]
                                            [--age-tolerance]

Arguments:
    --prefix                   str    Filename prefix for saved output files.
    --subdir                   str    Subdirectory where raw time series files are stored.
    --sex                      int    Sex filter. Accepts 1 (Male) or 2 (Female) only.
    --age-min                  float  Minimum participant age (inclusive).
    --age-max                  float  Maximum participant age (inclusive).
    --seed                     int    Random seed for reproducible male matching (default: 42).
    --match-to-female-metadata Path   Path to female_metadata.csv. When provided, selects
                                      males matched to that female set by site and age.
                                      Cannot be used with --sex.
    --age-tolerance            float  Maximum age difference in years when matching males
                                      to females (default: 2.0).
                                      Only used with --match-to-female-metadata.

Examples:
    python 03_build_connectivity_dataset.py
    Processes all available participants with no filtering.

    python 03_build_connectivity_dataset.py --prefix 1100 --subdir 1100_participants
    Builds the full dataset from files in the 1100_participants subdirectory.

To create an age matched male subset to the female subset, run in order:

    !python 02_download_participants
    !python 03_build_connectivity_dataset.py --sex 2 --prefix female
    !python 03_build_connectivity_dataset.py --match-to-female-metadata data/
    processed/female/female_metadata.csv
    
Processed datasets are saved to product/data/processed/
"""
!python 03_build_connectivity_dataset.py --subdir SDSU --prefix SDSU
```
